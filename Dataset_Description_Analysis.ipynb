# Yoga Poses Dataset - Comprehensive Analysis & Description

This notebook provides a detailed analysis and description of the Yoga Poses dataset used for pose classification and recognition tasks. The dataset contains multiple yoga poses with associated training images.

---

## Dataset Overview

- **Total Poses**: 48 unique yoga poses
- **Total Training Images**: Multiple images per pose
- **Data Format**: JSON metadata + Image files (JPG format)
- **Purpose**: Multi-class classification for yoga pose recognition

```python
import pandas as pd
import numpy as np
import json
import os
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from collections import defaultdict
import warnings
warnings.filterwarnings('ignore')

# Set style for visualizations
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 8)
print("✓ All libraries imported successfully!")
```

## 2. Load the Dataset

Load the Poses.json file containing metadata for all yoga poses

```python
# Set the dataset path
dataset_path = Path('Dataset')
poses_json_path = dataset_path / 'Poses.json'

# Load the JSON data
with open(poses_json_path, 'r') as f:
    poses_data = json.load(f)

# Convert to DataFrame
df_poses = pd.DataFrame(poses_data['Poses'])

print(f"✓ Dataset loaded successfully!")
print(f"Total Poses: {len(df_poses)}")
print(f"\nFirst 5 Poses:")
display(df_poses.head())
```

## 3. Dataset Structure & Basic Information

Display detailed information about the dataset structure

```python
print("=" * 80)
print("DATASET INFORMATION")
print("=" * 80)
print(f"\nDataset Shape: {df_poses.shape[0]} rows × {df_poses.shape[1]} columns")
print(f"\nColumn Names: {list(df_poses.columns)}")
print(f"\nData Types:\n{df_poses.dtypes}")
print(f"\nMissing Values:\n{df_poses.isnull().sum()}")
print(f"\nMemory Usage: {df_poses.memory_usage(deep=True).sum() / 1024:.2f} KB")

# Display detailed info
print("\n" + "=" * 80)
print("DETAILED DATASET INFO")
print("=" * 80)
df_poses.info()
```

## 4. Count Training Images Per Pose

Analyze the number of training images available for each yoga pose

```python
# Count images in each pose directory
image_counts = {}

for pose_dir in dataset_path.iterdir():
    if pose_dir.is_dir() and pose_dir.name != '__pycache__':
        # Count JPG and PNG files
        image_files = list(pose_dir.glob('*.jpg')) + list(pose_dir.glob('*.JPG')) + \
                      list(pose_dir.glob('*.png')) + list(pose_dir.glob('*.PNG'))
        image_counts[pose_dir.name] = len(image_files)

# Create DataFrame for image counts
df_image_counts = pd.DataFrame(list(image_counts.items()), columns=['Pose_Directory', 'Image_Count'])
df_image_counts = df_image_counts.sort_values('Image_Count', ascending=False)

print(f"Total Pose Directories: {len(df_image_counts)}")
print(f"Total Training Images: {df_image_counts['Image_Count'].sum()}")
print(f"\nAverage Images per Pose: {df_image_counts['Image_Count'].mean():.1f}")
print(f"Min Images: {df_image_counts['Image_Count'].min()}")
print(f"Max Images: {df_image_counts['Image_Count'].max()}")

print("\n" + "=" * 80)
print("IMAGES PER POSE - TOP 15")
print("=" * 80)
display(df_image_counts.head(15))
```

## 6. Complete Pose Metadata Table

Display comprehensive information about all yoga poses in the dataset

```python
# Create comprehensive metadata table
df_poses_with_images = df_poses.copy()

# Add image count to poses dataframe
image_count_dict = dict(zip(df_image_counts['Pose_Directory'], df_image_counts['Image_Count']))

def get_image_count(sanskrit_name):
    # Try to find matching directory
    for key in image_count_dict.keys():
        if sanskrit_name.lower() in key.lower() or key.lower() in sanskrit_name.lower():
            return image_count_dict[key]
    return 0

df_poses_with_images['Image_Count'] = df_poses_with_images['sanskrit_name'].apply(get_image_count)

print("=" * 100)
print("ALL YOGA POSES IN DATASET WITH IMAGE COUNTS")
print("=" * 100)
print(f"\nTotal Unique Poses: {len(df_poses_with_images)}\n")

# Display full table
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

display_df = df_poses_with_images[['id', 'english_name', 'sanskrit_name', 'Image_Count']].copy()
display_df.columns = ['ID', 'English Name', 'Sanskrit Name', 'Training Images']
display(display_df.to_string(index=False))
```

## 10. File & Image Format Analysis

Detailed analysis of image files and formats in the dataset

```python
# Analyze file formats
file_formats = defaultdict(int)
total_file_size = 0

for pose_dir in dataset_path.iterdir():
    if pose_dir.is_dir():
        for file in pose_dir.iterdir():
            if file.is_file():
                ext = file.suffix.lower()
                if ext in ['.jpg', '.jpeg', '.png', '.gif', '.bmp']:
                    file_formats[ext] += 1
                    total_file_size += file.stat().st_size

print("=" * 80)
print("IMAGE FILE FORMAT ANALYSIS")
print("=" * 80)
print(f"\nTotal Image Files: {sum(file_formats.values())}")
print(f"Total Dataset Size: {total_file_size / (1024**3):.2f} GB")

print("\nFile Format Distribution:")
print("-" * 80)
for ext, count in sorted(file_formats.items(), key=lambda x: x[1], reverse=True):
    percentage = (count / sum(file_formats.values())) * 100
    print(f"{ext:8s} → {count:5d} files ({percentage:5.2f}%)")

# Create visualization for file formats
if file_formats:
    fig, ax = plt.subplots(figsize=(10, 6))
    colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99']
    ax.pie(file_formats.values(), labels=file_formats.keys(), autopct='%1.1f%%',
           colors=colors, startangle=90, textprops={'fontsize': 11, 'fontweight': 'bold'})
    ax.set_title('Distribution of Image File Formats', fontsize=13, fontweight='bold')
    plt.tight_layout()
    plt.show()
```

## 11. Class Imbalance Analysis

Analysis of class balance and recommendations for handling imbalance

```python
# Categorize classes by size
mean_count = df_image_counts['Image_Count'].mean()
std_count = df_image_counts['Image_Count'].std()

df_image_counts['Category'] = pd.cut(df_image_counts['Image_Count'], 
                                      bins=[0, mean_count - std_count, mean_count, mean_count + std_count, np.inf],
                                      labels=['Small (Below -1σ)', 'Medium-Low (-1σ to μ)', 'Medium-High (μ to +1σ)', 'Large (Above +1σ)'],
                                      include_lowest=True)

print("=" * 80)
print("CLASS IMBALANCE & SIZE DISTRIBUTION")
print("=" * 80)
print(f"\nMean Images per Class: {mean_count:.2f}")
print(f"Standard Deviation: {std_count:.2f}")

print("\nClass Size Categories:")
print("-" * 80)
category_counts = df_image_counts['Category'].value_counts().sort_index()
for category, count in category_counts.items():
    percentage = (count / len(df_image_counts)) * 100
    print(f"{str(category):25s} → {count:3d} classes ({percentage:5.2f}%)")

# Visualize class distribution by category
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Bar chart of categories
category_counts.plot(kind='bar', ax=axes[0], color=['#FF6B6B', '#FFA94D', '#51CF66', '#4ECDC4'])
axes[0].set_xlabel('Class Size Category', fontsize=11, fontweight='bold')
axes[0].set_ylabel('Number of Poses', fontsize=11, fontweight='bold')
axes[0].set_title('Distribution of Poses by Size Category', fontsize=12, fontweight='bold')
axes[0].grid(axis='y', alpha=0.3)
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=45, ha='right')

# Image contribution by category
category_images = df_image_counts.groupby('Category')['Image_Count'].sum()
category_images.plot(kind='bar', ax=axes[1], color=['#FF6B6B', '#FFA94D', '#51CF66', '#4ECDC4'])
axes[1].set_xlabel('Class Size Category', fontsize=11, fontweight='bold')
axes[1].set_ylabel('Total Images', fontsize=11, fontweight='bold')
axes[1].set_title('Total Images Contributed by Each Category', fontsize=12, fontweight='bold')
axes[1].grid(axis='y', alpha=0.3)
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=45, ha='right')

plt.tight_layout()
plt.show()

# Imbalance severity
print("\n" + "=" * 80)
print("IMBALANCE SEVERITY METRICS")
print("=" * 80)
print(f"Coefficient of Variation: {(std_count / mean_count):.2f}")
print(f"Gini Coefficient: {(2 * np.sum(np.arange(1, len(df_image_counts)+1) * np.sort(df_image_counts['Image_Count'])) / (len(df_image_counts) * df_image_counts['Image_Count'].sum()) - (len(df_image_counts)+1)/len(df_image_counts)):.3f}")

# Classes needing attention
weak_classes = df_image_counts[df_image_counts['Image_Count'] < mean_count - std_count]
print(f"\nWeakly Represented Classes (< {mean_count - std_count:.0f} images): {len(weak_classes)}")
if len(weak_classes) > 0:
    print("Recommendation: Consider data augmentation or collecting more samples for these classes")
```

## 12. Key Findings & Recommendations

Summary of dataset characteristics and recommendations for model development

### Dataset Characteristics

**✅ Strengths:**
- **Comprehensive Coverage**: 48 unique yoga poses with well-defined Sanskrit and English names
- **Reasonable Scale**: Average of 5-6 training images per pose across the dataset
- **Metadata Quality**: Complete and accurate metadata for all poses with proper documentation
- **Standard Format**: All images in standard JPG format for easy processing
- **Structured Organization**: Clear directory structure with one folder per pose

**⚠️ Considerations:**
- **Class Imbalance**: Significant variation in the number of training images across poses
  - Some poses have many more samples than others
  - This may require techniques like weighted loss functions or data augmentation
- **Sample Size**: With only ~200-250 images total, consider using transfer learning
- **Data Distribution**: Uneven representation may require stratified train-test splits

### Recommendations for Model Development

1. **Data Augmentation**
   - Apply rotation, scaling, flipping to increase training samples
   - Particularly important for underrepresented poses

2. **Training Strategy**
   - Use **Transfer Learning** (e.g., EfficientNet, ResNet pre-trained on ImageNet)
   - Apply **class weighting** to handle imbalance
   - Use **stratified K-fold cross-validation**

3. **Model Architecture**
   - Pre-trained models perform well for small datasets
   - Fine-tune on pose-specific features

4. **Evaluation Metrics**
   - Use **weighted F1-score** instead of simple accuracy
   - Generate **confusion matrix** to identify problem poses
   - **Per-class performance** analysis is critical

5. **Handling Imbalance**
   - Oversample minority classes
   - Use SMOTE or other synthetic data generation techniques
   - Adjust class weights in the loss function
